```python
# Instalar ipyfilechooser si no está instalado
try:
    from ipyfilechooser import FileChooser
except ImportError:
    import sys
    !{sys.executable} -m pip install ipyfilechooser -q
    from ipyfilechooser import FileChooser

# Instalar opencv-python si no está instalado
try:
    import cv2
except ImportError:
    import sys
    !{sys.executable} -m pip install opencv-python -q
    import cv2

import ipywidgets as widgets
import os

# Inicializar el selector de archivos
fc = FileChooser(os.getcwd())
fc.title = '<b>Selecciona la carpeta de imágenes</b>'
fc.show_only_dirs = True

# Botón para cargar
btn_load = widgets.Button(
    description='Cargar Imágenes',
    button_style='success',
    icon='check'
)

# Output para mostrar mensajes
out = widgets.Output()

# Variable global para almacenar las imágenes
loaded_images = []

def load_images(b):
    global loaded_images
    folder_path = fc.selected_path
    image_extensions = ('.png', '.jpg', '.jpeg', '.gif', '.bmp', '.tiff', '.webp')
    loaded_images = []
    
    with out:
        out.clear_output()
        if not folder_path:
            print("⚠️ Por favor selecciona una carpeta.")
            return
        
        print(f"📂 Carpeta seleccionada: {folder_path}")
        if not os.path.exists(folder_path):
            print("❌ La carpeta no existe.")
            return

        print("⏳ Cargando imágenes...")
        count = 0
        for filename in os.listdir(folder_path):
            file_path = os.path.join(folder_path, filename)
            if os.path.isfile(file_path) and filename.lower().endswith(image_extensions):
                img = cv2.imread(file_path, cv2.IMREAD_COLOR)
                if img is not None:
                    loaded_images.append(img)
                    count += 1
        
        print(f"✅ Se cargaron {count} imágenes correctamente en la variable 'loaded_images'.")

btn_load.on_click(load_images)
display(widgets.VBox([fc, btn_load, out]))
```
