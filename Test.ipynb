```python
import pygame
import sys

# Inicializar pygame
pygame.init()

# Configuración de la ventana
WIDTH, HEIGHT = 800, 600
WINDOW_COLOR = (30, 30, 30)  # Fondo gris oscuro
BALL_COLOR = (255, 160, 13)   # Amarillo
BALL_RADIUS = 30

screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Pelota amarilla rebotando")

# Posición inicial de la pelota (centro)
ball_x = WIDTH // 2
ball_y = HEIGHT // 2

# Velocidad inicial de la pelota
vel_x = 5
vel_y = 4

clock = pygame.time.Clock()

running = True
while running:
    # Manejar eventos
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
        elif event.type == pygame.KEYDOWN and event.key == pygame.K_SPACE:
            running = False

    if not running:
        break

    # Actualizar posición de la pelota
    ball_x += vel_x
    ball_y += vel_y

    # Rebotar en los bordes horizontales
    if ball_x - BALL_RADIUS <= 0 or ball_x + BALL_RADIUS >= WIDTH:
        vel_x *= -1

    # Rebotar en los bordes verticales
    if ball_y - BALL_RADIUS <= 0 or ball_y + BALL_RADIUS >= HEIGHT:
        vel_y *= -1

    # Dibujar fondo y pelota
    screen.fill(WINDOW_COLOR)
    pygame.draw.circle(screen, BALL_COLOR, (ball_x, ball_y), BALL_RADIUS)

    # Actualizar la pantalla
    pygame.display.flip()

    # Limitar FPS
    clock.tick(60)
```
